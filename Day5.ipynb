```python
#Resampling 和 Groupby 时间分组
import pandas as pd

df = pd.DataFrame({
    "Timestamp": [
        "2026/11/01 08:00:00",
        "2026/11/01 09:00:00",
        "2026/11/01 10:00:00"
    ],
    "Temperature": [65.0, 65.5, 66.0]
})

##将时间字符串转换为 Pandas 可以识别和计算的时间格式
df["Timestamp"] = pd.to_datetime(df["Timestamp"])

#按小时分组
df = df.set_index("Timestamp")#将时间列设置为 DataFrame 的“索引（Index）”

hourly_avg = df["Temperature"].resample("h").mean()

print(hourly_avg)

#按天计算平均温度
daily_avg = df["Temperature"].resample("D").mean()
print(daily_avg)

#同时计算每天最高、最低和平均温度
daily_summary = df["Temperature"].resample("D").agg(
    ["mean", "max", "min", "std"]
)
print(daily_summary)
```

```python
#按设备和日期分组
import pandas as pd

df = pd.DataFrame({"Tool_ID":["A","B","C"],
    "Timestamp": [
        "2026/11/01 08:00:00",
        "2026/11/01 09:00:00",
        "2026/11/01 10:00:00"
    ],
    "Temperature": [65.0, 65.5, 66.0]
})

##将时间字符串转换为 Pandas 可以识别和计算的时间格式
df["Timestamp"] = pd.to_datetime(df["Timestamp"])

#按设备和日期分组
df["Date"] = df["Timestamp"].dt.date

daily_tool_avg = df.groupby(
    ["Tool_ID", "Date"]
)["Temperature"].mean()

print(daily_tool_avg)
```

```python
#移动平均 Rolling Mean(OR Moving Average)，作用是：计算一段时间窗口内的平均值，减少短期波动对趋势判断的干扰
import pandas as pd

df = pd.DataFrame({
    "Timestamp": [
        "08:00:",
        "09:00",
        "10:00",
        "11:00",
        "12:00"
    ],
    "Temperature": [65.0, 66.0, 68.0,70.0,73.0]
})

df["Rolling_Mean"] = (
    df["Temperature"]
    .rolling(window=3)
    .mean()
)

#print(df["Rolling_Mean"])

#计算温度的变化，下一条减去上一条
df["Temperature_Change"] = df["Temperature"].diff()
print(df["Temperature_Change"])
```
