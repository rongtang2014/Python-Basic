```python
import pandas as pd

#第一步：读取设备数据
df = pd.read_csv("equipment_data.csv")

#检查数据规模
df.shape

#第二步：检查数据是否完整
df.isna().sum()

#第三步：检查重复记录
df.duplicated().sum()
#df.duplicated(subset=['Tool_ID','Wafer_ID']).sum)

#缺失值和重复值处理
df_clean = df.drop_duplicates() #这会删除所有列完全相同的重复行，并把结果保存到新变量 df_clean

#第四步：计算每台设备的平均温度
temperature_summary = df.groupby("Tool_ID")["Temperature"].mean()

#统计每台设备的数据数量
df.groupby("Tool_ID")["Temperature"].count()

defect_mean = df.groupby("Tool_ID")["Defect_Count"].mean()
defect_mean.max() #找最大值
defect_mean.idxmax() #找最大数值对应的索引标签

#同时找出设备和最大缺陷数
highest_tool = defect_mean.idxmax()

highest_defect = defect_mean.max()

print(highest_tool)
print(highest_defect)

#生成每台设备的综合统计表
summary = df.groupby("Tool_ID").agg({
    "Temperature": ["mean", "max", "count"],
    "Defect_Count": ["mean", "max", "count"]
})

#从汇总表中找出缺陷最高的设备
highest_tool = summary[("Defect_Count", "mean")].idxmax()

#提取这台设备的完整统计信息
print(summary.loc[highest_tool])

#保存汇总结果
summary.to_csv("equipment_summary.csv", index=True)

#把索引恢复成普通列
summary = summary.reset_index()

#筛选汇总表（找出平均缺陷数大于20的设备）
high_defect_tools = summary[                # 第一个 summary：接收筛选条件，筛选整行
    summary[("Defect_Count", "mean")] > 20  # 第二个 summary：读取温度列，生成条件
]
```
